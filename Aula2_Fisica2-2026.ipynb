## Cinemática de rotações, Energia Rotacional e Momento de Inércia

A tabela abaixo traz um resumo das relações entre as **grandezas angulares** (rotações) com suas respectivas **grandezas lineares** vistas até agora.

Translação pura (1D) | Rotação pura (eixo Fixo)       | Relação    
:--------------------|:-------------------------------|:--------
Posição x            |  Posição angular θ             | s = rθ  (*sempre em radianos*!!)
Velocidade v = dx/dt |  Velocidade angular ω = dθ/dt  | v = ωr
Aceleração a = dv/dt |  Aceleração angular $\alpha$ = dω/dt  | a$_t$ = $\alpha$ r; <br>a$_c = \frac{v^2}{r} = \omega^2 r$
Massa      m         |  Momento de inércia I          | (veja abaixo)

As três primeiras estabelecem um "mapa" direto entre os conceitos de posição, velocidade e aceleração, para ambas as situações, permitindo aproveitar a experiência acumulada da cinemática do movimento linear. É interessante notar, porém, a **novidade no conceito associado à massa** (inércia) do corpo. Para entender melhor como isso surge, é conveniente usar a energia cinética associada à rotação de um conjunto de partículas, expressa em termos das variáveis angulares (do mov. de rotação).

Seja um uma partícula de massa $m$, em movimento circular de velocidade $\omega$ a uma distância $\rho$ do eixo. Neste caso, a energia cinética da partícula é dada por $K = \frac{1}{2} m v^2 = \frac{1}{2}m \rho^2 \omega^2 $. Se considerarmos agora um sistema de N partículas, por extensão, teremos que a **energia cinética de rotação** do sistema é dada pela soma das partes:

$$K = \frac{1}{2} (\sum_{i=1}^N m_i \rho_i^2) \, \omega_i^2 = \frac{1}{2} I \omega_i^2 $$

onde $I= \sum_{i=1}^N m_i \,\rho_i^2$ representa a grandeza denominada de **momento de inércia** do sistema.


O momento de inércica tem um papel análogo à massa, no sentido de representar a "**inércia**" de rotação de um corpo ou sistema de partículas.

<br>
Nos casos mais gerais, que podem incluir tanto sistemas de partículas discretas como corpos rígidos com distribuições contínuas de massa, teremos:

$$I = \sum_{i=1}^N \rho_i^2 \, m_i \,\, \textrm{   (caso discreto)}$$
$$I = \int \rho^2 dm  \,\, \textrm{   (caso contínuo)} $$  

Nesta aula exploramos alguns exemplos simples de como calcular o momento de inércia para diferetentes distribuições de massa. d) $\frac{1}{3}ML^2$  <br><br>


#### Resolução:
<center><img src=".\figuras\barra_solucao.png"/></center>

\begin{align}
dm & = \frac{M}{L}dx \\
I_{CM} = \int_{x=-L/2}^{x=+L/2} x^2 dm & = \frac{M}{L} \int_{-L/2}^{L/2} x^2 dx \\
I_{CM} = \frac{M}{3L}\,\left[ \left( \frac{L}{2} \right)^3 - \left( -\frac{L}{2} \right)^3 \right]
     & = \frac{1}{12} M L^2
\end{align}

### <font color="blue">Teorema dos eixos paralelos (Steiner)</font>
Um resultado bastante útil, conhecido como teorema de Steiner, permite calcular o momento de inércia $I$ de massa $M$, em relação a um eixo que é paralelo a um eixo que passa pelo CM (centro de massa). Se a distância entre os eixos for $h$, a relação entre $I$ e $I_{CM}$ é dada por:

$$I = I_{CM} + Mh^2$$

Para praticar, use esse resultado para calcular o momenot de inércia de uma barra girando ao redor de um eixo perpendicular à barra a uma de suas extremidades... (depois confira com o cálculo direto, a partir da definição, que você fez no item a)


$$I = I_{CM} + M(L/2)^2 = ML^2 \cdot \left( \frac{1}{12}+\frac{1}{4} \right) = \frac{1+3}{12} ML^2 = \frac{1}{3} ML^2 $$

### <font color="blue">Teorema dos eixos perpendiculares</font>

Há um resultado bem particular para corpos planos e delgados, mas que pode ser útil em algumas situações. É chamado de teorema dos eixos perpendiculares e diz respeito a situação descrita abaixo.

<center><img src=".\figuras\DiscoPlanoXYZ.png" width="300" /></center>

$$ I_x = \int y^2 dm $$
$$ I_y = \int x^2 dm $$
$$ I_z = \int \rho^2 dm = \int (x^2+y^2) \,dm $$

Note que $I_z = I_x + I_y$, isto é, a soma dos momentos de inércia dos dois eixos perpendiculares no plano é igual ao momento de inércia com relação ao terceiro eixo (perpendicular ao plano do disco), passando pela intersecção do outros dois.

No caso do Exemplo 3, devido a simetria, $I_x = I_y = \frac{1}{4}MR^2$.
